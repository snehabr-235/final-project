```python
categorical_columns = [
    'product',
    'category',
    'payment_method',
    'city',
    'state',
    'customer_type'
]

for col in categorical_columns:
    df[col] = df[col].fillna('Unknown')

print("Categorical missing values handled successfully!")
```

```python

# Correct inconsistent quantity entries

df['quantity'] = df['quantity'].replace({
    'two': '2',
    '3 pcs': '3',
    'unknown': np.nan
})

# Convert quantity into numeric format

df['quantity'] = pd.to_numeric(
    df['quantity'], errors='coerce'
)

# Identify invalid zero or negative quantities

df.loc[df['quantity'] <= 0, 'quantity'] = np.nan

print("Quantity column cleaned successfully!")

print(df['quantity'].value_counts(dropna=False).sort_index())
```

```python
# Replace unusually high quantities with missing values

df.loc[df['quantity'] > 5, 'quantity'] = np.nan

# Calculate median quantity

quantity_median = df['quantity'].median()

print("Quantity Median:", quantity_median)

# Fill missing quantities using median

df['quantity'] = df['quantity'].fillna(quantity_median)

print("Missing quantities handled successfully!")

print(df['quantity'].value_counts().sort_index())
```

```python

# Remove currency symbols and commas

df['unit_price'] = (
    df['unit_price']
    .astype('string')
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)

# Convert unit price into numeric format

df['unit_price'] = pd.to_numeric(
    df['unit_price'], errors='coerce'
)

print("Unit price converted successfully!")

print(df['unit_price'].describe())
```

```python

print("Negative prices:")
print(df[df['unit_price'] < 0]['unit_price'].value_counts())

print("\nVery high prices:")
print(df[df['unit_price'] > 10000]['unit_price'].value_counts())

print("\nPrice distribution:")
print(df['unit_price'].sort_values(ascending=False).head(15))
```

```python

# Replace negative and unusually high prices with NaN

df.loc[
    (df['unit_price'] < 0) |
    (df['unit_price'] > 10000),
    'unit_price'
] = np.nan

print("Invalid unit prices handled successfully!")

print(df['unit_price'].describe())
print("\nMissing unit prices:", df['unit_price'].isnull().sum())
```

```python
unit_price_median = df['unit_price'].median()

df['unit_price'] = df['unit_price'].fillna(unit_price_median)

print("Missing unit prices filled successfully!")
print("Remaining missing values:", df['unit_price'].isnull().sum())
```

```python

# Convert customer_age into numeric format

df['customer_age'] = pd.to_numeric(df['customer_age'], errors='coerce')

print("Customer age converted successfully!")
print(df['customer_age'].dtype)
```

```python

# Fill missing customer ages using median

age_median = df['customer_age'].median()

df['customer_age'] = df['customer_age'].fillna(age_median)

print("Missing customer ages filled successfully!")
print("Remaining missing values:", df['customer_age'].isnull().sum())
```

```python

# Convert discount_percent into numeric format

df['discount_percent'] = pd.to_numeric(df['discount_percent'], errors='coerce')

print("Discount percentage converted successfully!")
print(df['discount_percent'].dtype)
```

```python

# Fill missing discount percentages using median

discount_median = df['discount_percent'].median()

df['discount_percent'] = df['discount_percent'].fillna(discount_median)

print("Missing discount percentages filled successfully!")
print("Remaining missing values:", df['discount_percent'].isnull().sum())
```

```python

# Convert order_date into datetime format

date_series = df['order_date'].astype('string').str.strip()

df['order_date'] = pd.NaT

# DD/MM/YYYY
mask = date_series.str.match(r'^\d{1,2}/\d{1,2}/\d{4}$', na=False)
df.loc[mask, 'order_date'] = pd.to_datetime(
    date_series[mask], format='%d/%m/%Y', errors='coerce'
)

# YYYY-MM-DD
mask = date_series.str.match(r'^\d{4}-\d{1,2}-\d{1,2}$', na=False)
df.loc[mask, 'order_date'] = pd.to_datetime(
    date_series[mask], format='%Y-%m-%d', errors='coerce'
)

# YYYY/MM/DD
mask = date_series.str.match(r'^\d{4}/\d{1,2}/\d{1,2}$', na=False)
df.loc[mask, 'order_date'] = pd.to_datetime(
    date_series[mask], format='%Y/%m/%d', errors='coerce'
)

# MM-DD-YYYY
mask = date_series.str.match(r'^\d{1,2}-\d{1,2}-\d{4}$', na=False)
df.loc[mask, 'order_date'] = pd.to_datetime(
    date_series[mask], format='%m-%d-%Y', errors='coerce'
)

# DD Month YYYY
mask = date_series.str.match(
    r'^\d{1,2}\s+[A-Za-z]{3,9}\s+\d{4}$', na=False
)
df.loc[mask, 'order_date'] = pd.to_datetime(
    date_series[mask], format='mixed',
    dayfirst=True, errors='coerce'
)

print("Order dates converted successfully!")
print("Valid dates:", df['order_date'].notna().sum())
print("Missing dates:", df['order_date'].isna().sum())
```

```python

# Calculate total product value

df['total_product_value'] = df['quantity'] * df['unit_price']

print("Total product value column created successfully!")
print(df[['quantity', 'unit_price', 'total_product_value']].head())
```

```python

# Calculate final amount after discount and shipping

df['final_calculated_amount'] = (
    df['total_product_value']
    - df['discount_amount']
    + df['shipping_fee']
)

print("Final calculated amount column created successfully!")
print(df[['total_product_value', 'discount_amount', 'shipping_fee', 'final_calculated_amount']].head())
```
